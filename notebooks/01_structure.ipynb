```python
import ase.io
import numpy as np
file_path= "/Users/kasperhuysentruyt/Documents/leven /Cusp/cusp-mlff-gcmc/data/IRMOF-1.cif"
atoms = ase.io.read(file_path, format="cif")
print(atoms)
print(len(atoms))
print(atoms.cell)
print(np.sum(atoms.get_masses()))
print(atoms.get_chemical_symbols())
print(atoms.get_distances(0, 4, mic=True))
print(atoms.get_chemical_symbols()[0])
list_of_distances = []
for i in range(len(atoms)):
    if atoms.get_chemical_symbols()[i] == "Zn":
        for j in range(len(atoms)):
            if atoms.get_chemical_symbols()[j] == "O":
                list_of_distances.append(atoms.get_distances(i, j, mic=True))
min_distance = min(list_of_distances)
where_min = np.where(np.array(list_of_distances) == min_distance)
print(np.sum(where_min))
print("Minimum distance between Zn and O atoms:", min_distance)
elements = atoms.get_chemical_symbols()
print("Elements in the structure:", set(elements))

        
```
